```python
# Tic-Tac-Toe
# HUMAN = O
# COMPUTER = X

# 1. Create a 3 x 3 board
board = [" " for _ in range(9)]


# Function to display the board
def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


# Function to check whether a player has won
def check_winner(player):
    winning_positions = [
        (0, 1, 2),  # Row 1
        (3, 4, 5),  # Row 2
        (6, 7, 8),  # Row 3
        (0, 3, 6),  # Column 1
        (1, 4, 7),  # Column 2
        (2, 5, 8),  # Column 3
        (0, 4, 8),  # Diagonal
        (2, 4, 6)   # Diagonal
    ]

    for a, b, c in winning_positions:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


# Function to check whether board is full
def board_full():
    return " " not in board


# 2. Start the game
print("TIC-TAC-TOE")
print("Human = O")
print("Computer = X")

while True:

    # -------------------------
    # HUMAN'S TURN
    # -------------------------
    display_board()

    position = int(input("Enter position (1-9): ")) - 1

    # Check if position is valid
    if position < 0 or position > 8:
        print("Invalid position!")
        continue

    if board[position] != " ":
        print("Cell already occupied!")
        continue

    # Place O
    board[position] = "O"

    # Check HUMAN win
    if check_winner("O"):
        display_board()
        print("HUMAN WINS!")
        break

    # Check draw
    if board_full():
        display_board()
        print("DRAW!")
        break


    # -------------------------
    # COMPUTER'S TURN
    # -------------------------

    # Select the first available cell
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            break

    print("Computer played:", i + 1)

    # Check COMPUTER win
    if check_winner("X"):
        display_board()
        print("COMPUTER WINS!")
        break

    # Check draw
    if board_full():
        display_board()
        print("DRAW!")
        break
```
